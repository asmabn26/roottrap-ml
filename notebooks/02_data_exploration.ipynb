```python
import pandas as pd

df_cicids = pd.read_csv('../data/raw/cicids2017/cicids2017_cleaned.csv')
print("Shape :", df_cicids.shape)
print("\nColonnes :", df_cicids.columns.tolist())
```

```python
df_cicids.head()
```

```python
print("Valeurs manquantes :", df_cicids.isnull().sum().sum())
print("Taille en mémoire :", df_cicids.memory_usage(deep=True).sum() / 1024**2, "MB")
```

```python
import matplotlib.pyplot as plt

print("Distribution des labels :")
print(df_cicids['Attack Type'].value_counts())

plt.figure(figsize=(14,5))
df_cicids['Attack Type'].value_counts().plot(kind='bar', color='steelblue')
plt.title("Distribution des types d'attaques — CICIDS2017")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/raw/cicids2017/label_distribution.png')
plt.show()
```

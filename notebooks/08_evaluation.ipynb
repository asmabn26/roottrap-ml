```python
import joblib, numpy as np, json
from sklearn.metrics import f1_score, precision_score, recall_score, accuracy_score

X_test  = np.load('../data/processed/X_test.npy')
y_test  = np.load('../data/processed/y_test.npy')

with open('../data/label_encoding.json') as f:
    mapping = json.load(f)
class_names = list(mapping.keys())

rf = joblib.load('../models/random_forest_model.pkl')
y_pred_rf = rf.predict(X_test)

resultats = {
    'Random Forest': {
        'accuracy': accuracy_score(y_test, y_pred_rf),
        'precision': precision_score(y_test, y_pred_rf, average='weighted'),
        'recall':    recall_score(y_test, y_pred_rf, average='weighted'),
        'f1':        f1_score(y_test, y_pred_rf, average='weighted'),
    }
}

import pandas as pd
df_res = pd.DataFrame(resultats).T
print(df_res.round(4))
df_res.to_csv('../data/evaluation_results.csv')
```

```python
rapport = f"""
# Rapport d'évaluation — Sprint 2 RootTrap

## Random Forest
- Accuracy  : {resultats['Random Forest']['accuracy']:.4f}
- Precision : {resultats['Random Forest']['precision']:.4f}
- Recall    : {resultats['Random Forest']['recall']:.4f}
- F1-score  : {resultats['Random Forest']['f1']:.4f}

## Observations
- Dataset : CICIDS2017 ({X_test.shape[0]} échantillons test)
- Classes  : {class_names}
- SMOTE appliqué pour corriger le déséquilibre (83% Normal Traffic)

## Prochaines étapes Sprint 3
- Ajouter DBSCAN pour le clustering comportemental
- Intégrer les logs Cowrie de P2
- Connecter l'API Flask vers P4
"""

with open('../docs/evaluation_report.md', 'w', encoding='utf-8') as f:
    f.write(rapport)

print("Rapport sauvegardé : docs/evaluation_report.md")
```

```python
import json
import pandas as pd

# Charger le JSON MITRE
with open('../data/mitre/enterprise-attack.json', 'r') as f:
    data = json.load(f)

# Extraire les objets de type "attack-pattern" (= techniques)
techniques = [obj for obj in data['objects'] if obj['type'] == 'attack-pattern']
print(f"Nombre de techniques : {len(techniques)}")

# Construire un DataFrame
records = []
for t in techniques:
    tactic = [p['phase_name'] for p in t.get('kill_chain_phases', [])]
    records.append({
        'technique_id': t.get('external_references', [{}])[0].get('external_id', ''),
        'name': t.get('name', ''),
        'tactics': tactic,
        'description': t.get('description', '')[:200]
    })

df = pd.DataFrame(records)
print(df.head(10))
print("\nTactiques uniques :")
print(df['tactics'].explode().value_counts())
```

```python
import matplotlib.pyplot as plt

tactic_counts = df['tactics'].explode().value_counts()

plt.figure(figsize=(12,6))
tactic_counts.plot(kind='bar')
plt.title("Répartition des techniques par tactique ATT&CK")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df.to_csv('../data/mitre/techniques_reference.csv', index=False)
print("Dataset sauvegardé")
```

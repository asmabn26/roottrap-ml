```python
import numpy as np
import joblib
import json
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (classification_report, confusion_matrix,
                             f1_score, precision_score, recall_score,
                             accuracy_score)

X_test  = np.load('../data/processed/X_test.npy')
y_test  = np.load('../data/processed/y_test.npy')

with open('../data/label_encoding.json') as f:
    mapping = json.load(f)
class_names = list(mapping.keys())

rf      = joblib.load('../models/random_forest_model.pkl')
y_pred  = rf.predict(X_test)

print("=" * 50)
print("RAPPORT D'ÉVALUATION — RANDOM FOREST")
print("=" * 50)
print(classification_report(y_test, y_pred, target_names=class_names))
```

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix — Random Forest Final")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/final_confusion_matrix.png', dpi=150)
plt.show()
```

```python
accuracy  = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall    = recall_score(y_test, y_pred, average='weighted')
f1        = f1_score(y_test, y_pred, average='weighted')

rapport = f"""
# Rapport Final ML — RootTrap Sprint 3

## Métriques Globales — Random Forest
| Métrique  | Score  |
|-----------|--------|
| Accuracy  | {accuracy:.4f} |
| Precision | {precision:.4f} |
| Recall    | {recall:.4f} |
| F1-score  | {f1:.4f} |

## Dataset
- Source       : CICIDS2017 (cleaned)
- Lignes test  : {X_test.shape[0]}
- Features     : {X_test.shape[1]}
- Classes      : {class_names}

## Modèles entraînés
- Random Forest : models/random_forest_model.pkl
- LSTM          : models/lstm_model.h5
- Scaler        : models/scaler.pkl

## Fichiers générés
- Mapper ATT&CK étendu  : data/attck_extended_mapping.json
- Kill Chain DAG        : data/kill_chain_dag.png
- Confusion Matrix      : data/final_confusion_matrix.png

## Prochaines étapes — Sprint 4
- Fine-tuning Grid Search
- Analyse malware patterns
- Tests adversarial attacks
- Revue code sécurité P4
"""

with open('../docs/final_evaluation_report.md', 'w', encoding='utf-8') as f:
    f.write(rapport)

print(rapport)
print("✅ Rapport sauvegardé : docs/final_evaluation_report.md")
```
